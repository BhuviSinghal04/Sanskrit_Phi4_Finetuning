```python
# INSPECTING DATASET
# -------------------

print("=" * 70)
print("First 5 Rows")
print("=" * 70)

display(df.head())

print("\n" + "=" * 70)
print("Missing Values")
print("=" * 70)

display(df.isnull().sum())

print("\n" + "=" * 70)
print("Duplicate Rows")
print("=" * 70)

print(df.duplicated().sum())

print("\n" + "=" * 70)
print("Meter Distribution (Top 15)")
print("=" * 70)

display(df["meter"].value_counts().head(15))
```

```python
# REMOVING DUPLICATE ROWS
# -------------------------

initial_rows = len(df)

df = df.drop_duplicates().reset_index(drop=True)

removed = initial_rows - len(df)

print("=" * 70)
print("Duplicate Removal Completed")
print("=" * 70)

print(f"Initial Rows   : {initial_rows}")
print(f"Removed Rows   : {removed}")
print(f"Remaining Rows : {len(df)}")

# Save checkpoint
checkpoint_path = OUTPUT_DIR / "01_dataset_without_duplicates.csv"
df.to_csv(checkpoint_path, index=False)

print("\n✅ Checkpoint saved to:")
print(checkpoint_path)
```

```python
# Remove Verses Containing "उवाच"
# --------------------------------


initial_rows = len(df)

mask = df["sa"].str.contains("उवाच", na=False)

removed = mask.sum()

df = df[~mask].reset_index(drop=True)

print("=" * 70)
print("Removed Verses Containing 'उवाच'")
print("=" * 70)

print(f"Initial Rows   : {initial_rows}")
print(f"Removed Rows   : {removed}")
print(f"Remaining Rows : {len(df)}")

# Save checkpoint
checkpoint_path = OUTPUT_DIR / "02_dataset_without_uvacha.csv"
df.to_csv(checkpoint_path, index=False)

print("\n✅ Checkpoint saved to:")
print(checkpoint_path)
```

```python
# INITIALIZE SKRUTABLE METER VALIDATOR
# -------------------------------------------

from skrutable.meter_identification import MeterIdentifier


mi = MeterIdentifier()


sample_verse = df.loc[0, "sa"]

print("=" * 70)
print("Sample Sanskrit Verse")
print("=" * 70)
print(sample_verse)

print("\n" + "=" * 70)
print("Running Skrutable Meter Identification...")
print("=" * 70)

result = mi.identify_meter(
    sample_verse,
    from_scheme="DEV",
    resplit_option="resplit_max"
)

print("Meter Label:")
print(result.meter_label)

print("\nSyllable Pattern:")
print(result.syllable_weights)

print("\nTotal Syllables:")
print(len(result.syllable_weights.replace("\n", "")))
```

```python
# CHECK DUPLICATE COLUMNS

from collections import Counter

counts = Counter(df.columns)

duplicates = {k: v for k, v in counts.items() if v > 1}

print("Duplicate columns:")
print(duplicates)
```

```python
# Verify Regex on Sample Labels

import re

VALID_LABEL_REGEX = re.compile(
    r"^anuṣṭubh \(\d,\d: (?!asamīcīna).+?[;,] \d,\d: (?!asamīcīna).+?\)$"
)

sample_labels = []

for verse in df["sa"].head(5):
    meter = mi.identify_meter(
        verse,
        from_scheme="DEV",
        resplit_option="resplit_max"
    )
    sample_labels.append(meter.meter_label)

print("=" * 70)
print("Regex Verification")
print("=" * 70)

for i, label in enumerate(sample_labels, 1):
    print(f"{i}. {label}")
    print("Match:", bool(VALID_LABEL_REGEX.match(label)))
    print()
```

```python

# Validate All Verses Using Skrutable
# ---------------------------------------

import pandas as pd
from tqdm.auto import tqdm

tqdm.pandas()

def analyze_verse(verse):
    try:
        meter = mi.identify_meter(
            verse,
            from_scheme="DEV",
            resplit_option="resplit_max"
        )

        label = meter.meter_label
        syllables = len(meter.syllable_weights.replace("\n", ""))

        return pd.Series({
            "meter_label": label,
            "syllable_count": syllables,
            "is_full_anustubh": bool(VALID_LABEL_REGEX.match(label)),
            "is_32_syllables": syllables == 32
        })

    except Exception:
        return pd.Series({
            "meter_label": "ERROR",
            "syllable_count": 0,
            "is_full_anustubh": False,
            "is_32_syllables": False
        })

print("=" * 70)
print("Validating Sanskrit verses...")
print("=" * 70)

results = df["sa"].progress_apply(analyze_verse)


df[
    [
        "meter_label",
        "syllable_count",
        "is_full_anustubh",
        "is_32_syllables"
    ]
] = results

print("\nValidation Complete")
print("=" * 70)

print(f"Full Anuṣṭubh : {int(df['is_full_anustubh'].sum())}")
print(f"32 Syllables  : {int(df['is_32_syllables'].sum())}")
print(f"Errors        : {int((df['meter_label'] == 'ERROR').sum())}")

checkpoint_path = OUTPUT_DIR / "03_validated_dataset.csv"

df.to_csv(checkpoint_path, index=False)

print("\nCheckpoint saved to:")
print(checkpoint_path)
```

```python
# Create Final Filtered Dataset
# -----------------------------

filtered_df = df[df["is_full_anustubh"]].reset_index(drop=True)

print("=" * 70)
print("Filtered Dataset Created")
print("=" * 70)

print(f"Original Rows : {len(df)}")
print(f"Filtered Rows : {len(filtered_df)}")
print(f"Removed Rows  : {len(df) - len(filtered_df)}")

filtered_path = OUTPUT_DIR / "04_filtered_dataset.csv"
filtered_df.to_csv(filtered_path, index=False)

print("\n✅ Saved to:")
print(filtered_path)
```

```python
#Create final_prompt
# ------------------

RULES = """
• The verse must follow the Anuṣṭubh meter.
• The verse must contain exactly 32 syllables.
• Output exactly four lines (one pāda per line).
• Use only Sanskrit in Devanagari script.
• Do not include explanations or extra text.
"""

def build_prompt(row):
    return f"""Objective:

Generate a Sanskrit verse in Anuṣṭubh meter based on a given meaning while strictly following the metrical rules.

Rules of Chandas:
{RULES}

Language:
The generated verse must be in Sanskrit in Devanagari script.
The input meaning is given in Hindi.
Do NOT include English words, transliteration, labels, or explanations.

Output Format:
(Output exactly 4 lines, one pada per line. Nothing else.)

Now generate the Sanskrit verse for the given Hindi meaning.

Meaning:
{row['hi']}

Sanskrit Verse:
"""

final_df["final_prompt"] = final_df.apply(build_prompt, axis=1)

print("=" * 70)
print("Final Prompt Column Created")
print("=" * 70)

print(final_df["final_prompt"].iloc[0])
```

```python
# ============================================================
# Task 2A - Load Final Dataset
# ============================================================

import pandas as pd
from pathlib import Path

PROJECT_DIR = Path("/content/drive/MyDrive/Sanskrit_Phi4_Finetuning")

DATASET_PATH = PROJECT_DIR / "outputs" / "06_final_dataset.csv"

df = pd.read_csv(DATASET_PATH)

print("=" * 70)
print("Dataset Loaded Successfully")
print("=" * 70)

print(f"Dataset Shape : {df.shape}")

print("\nColumns:")
print(df.columns.tolist())

display(df.head())
```

```python
# ============================================================
# Task 2B - Create Train / Validation / Test Split (90:5:5)
# ============================================================

import json
from pathlib import Path
from sklearn.model_selection import train_test_split

# ------------------------------------------------------------
# Reproducibility
# ------------------------------------------------------------
SEED = 42

# ------------------------------------------------------------
# Required Columns
# ------------------------------------------------------------
INSTRUCTION_COL = "final_prompt"
INPUT_COL = "hi"
OUTPUT_COL = "sa"

# Keep only required columns
dataset = df[[INSTRUCTION_COL, INPUT_COL, OUTPUT_COL]].copy()

# Rename columns for training
dataset.columns = ["instruction", "input", "output"]

# ------------------------------------------------------------
# Train (90%) | Temp (10%)
# ------------------------------------------------------------
train_df, temp_df = train_test_split(
    dataset,
    test_size=0.10,
    random_state=SEED,
    shuffle=True,
)

# ------------------------------------------------------------
# Validation (5%) | Test (5%)
# ------------------------------------------------------------
val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=SEED,
    shuffle=True,
)

# ------------------------------------------------------------
# Save Dataset Splits
# ------------------------------------------------------------
SPLIT_DIR = PROJECT_DIR / "outputs" / "dataset_splits"
SPLIT_DIR.mkdir(parents=True, exist_ok=True)

train_df.to_csv(SPLIT_DIR / "train.csv", index=False)
val_df.to_csv(SPLIT_DIR / "validation.csv", index=False)
test_df.to_csv(SPLIT_DIR / "test.csv", index=False)

# ------------------------------------------------------------
# Save Split Configuration
# ------------------------------------------------------------
split_config = {
    "seed": SEED,
    "train_ratio": 0.90,
    "validation_ratio": 0.05,
    "test_ratio": 0.05,
    "instruction_column": "instruction",
    "input_column": "input",
    "output_column": "output",
}

with open(SPLIT_DIR / "split_config.json", "w") as f:
    json.dump(split_config, f, indent=4)

# ------------------------------------------------------------
# Display Summary
# ------------------------------------------------------------
print("=" * 70)
print("Dataset Split Completed Successfully")
print("=" * 70)

print(f"Total Samples      : {len(dataset):,}")
print(f"Training Samples   : {len(train_df):,}")
print(f"Validation Samples : {len(val_df):,}")
print(f"Test Samples       : {len(test_df):,}")

print("\nFiles Saved To:")
print(SPLIT_DIR)
```

```python
# ============================================================
# Task 3A - Create Reproducible Configuration Files
# ============================================================

import json
from pathlib import Path

CONFIG_DIR = PROJECT_DIR / "outputs" / "config"
CONFIG_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# Seed Configuration
# ------------------------------------------------------------
seed_config = {
    "seed": 42
}

# ------------------------------------------------------------
# LoRA Configuration
# ------------------------------------------------------------
lora_config = {
    "r": 8,
    "lora_alpha": 16,
    "lora_dropout": 0.05,
    "bias": "none",
    "task_type": "CAUSAL_LM"
}

# ------------------------------------------------------------
# Training Configuration
# ------------------------------------------------------------
training_config = {
    "model_name": "microsoft/Phi-4-mini-instruct",
    "load_in_4bit": True,
    "max_seq_length": 2048,
    "train_batch_size": 2,
    "eval_batch_size": 2,
    "gradient_accumulation_steps": 8,
    "learning_rate": 2e-4,
    "weight_decay": 0.01,
    "num_train_epochs": 3,
    "warmup_ratio": 0.03,
    "lr_scheduler_type": "cosine",
    "logging_steps": 10,
    "evaluation_strategy": "steps",
    "save_strategy": "steps",
    "load_best_model_at_end": True,
    "metric_for_best_model": "eval_loss",
    "greater_is_better": False,
    "seed": 42
}

# ------------------------------------------------------------
# Save JSON Files
# ------------------------------------------------------------
with open(CONFIG_DIR / "seed_config.json", "w") as f:
    json.dump(seed_config, f, indent=4)

with open(CONFIG_DIR / "lora_config.json", "w") as f:
    json.dump(lora_config, f, indent=4)

with open(CONFIG_DIR / "training_config.json", "w") as f:
    json.dump(training_config, f, indent=4)

print("=" * 70)
print("Configuration Files Saved Successfully")
print("=" * 70)

print(f"Seed Config      : {CONFIG_DIR/'seed_config.json'}")
print(f"LoRA Config      : {CONFIG_DIR/'lora_config.json'}")
print(f"Training Config  : {CONFIG_DIR/'training_config.json'}")
```

```python
# ============================================================
# Task 3B - Environment Setup and Reproducibility
# ============================================================

import os
import random
import numpy as np
import torch

SEED = 42

# Python
random.seed(SEED)

# NumPy
np.random.seed(SEED)

# PyTorch
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed(SEED)
    torch.cuda.manual_seed_all(SEED)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

os.environ["PYTHONHASHSEED"] = str(SEED)

print("=" * 70)
print("Reproducibility Enabled")
print("=" * 70)

print("Seed :", SEED)
print("CUDA :", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU  :", torch.cuda.get_device_name(0))
```

```python
# ============================================================
# Task 3C - Load Phi-4 Mini Instruct (4-bit)
# ============================================================

import torch
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
)

MODEL_NAME = "microsoft/Phi-4-mini-instruct"

# ------------------------------------------------------------
# BitsAndBytes Configuration
# ------------------------------------------------------------
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.float16,
)

print("=" * 70)
print("Loading Tokenizer...")
print("=" * 70)

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

print("=" * 70)
print("Loading Model...")
print("=" * 70)

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=bnb_config,
    device_map="auto",
    torch_dtype=torch.float16,
    trust_remote_code=False,      # IMPORTANT
)

model.config.use_cache = False

print("=" * 70)
print("Model Loaded Successfully")
print("=" * 70)

print("Model :", MODEL_NAME)
print("Device:", next(model.parameters()).device)
print("Dtype :", next(model.parameters()).dtype)
```

```python
# ============================================================
# Task 3D - Configure LoRA for Phi-4 Mini Instruct
# ============================================================

from peft import LoraConfig, get_peft_model, TaskType

# ------------------------------------------------------------
# LoRA Configuration
# ------------------------------------------------------------
lora_config = LoraConfig(
    r=8,
    lora_alpha=16,
    lora_dropout=0.05,
    bias="none",
    task_type=TaskType.CAUSAL_LM,
    target_modules="all-linear",
)

print("=" * 70)
print("Applying LoRA...")
print("=" * 70)

model = get_peft_model(model, lora_config)

model.print_trainable_parameters()

print("=" * 70)
print("LoRA Applied Successfully")
print("=" * 70)
```

```python
# ============================================================
# Task 3E - Load and Prepare Training Dataset
# ============================================================

from pathlib import Path
import pandas as pd
from datasets import Dataset

# ------------------------------------------------------------
# Recreate Project Paths
# ------------------------------------------------------------
PROJECT_DIR = Path("/content/drive/MyDrive/Sanskrit_Phi4_Finetuning")
OUTPUT_DIR = PROJECT_DIR / "outputs"

# ------------------------------------------------------------
# Dataset Paths
# ------------------------------------------------------------
TRAIN_PATH = OUTPUT_DIR / "dataset_splits" / "train.csv"
VALID_PATH = OUTPUT_DIR / "dataset_splits" / "validation.csv"

print("Train Path :", TRAIN_PATH)
print("Valid Path :", VALID_PATH)

# ------------------------------------------------------------
# Verify Files Exist
# ------------------------------------------------------------
assert TRAIN_PATH.exists(), f"Missing: {TRAIN_PATH}"
assert VALID_PATH.exists(), f"Missing: {VALID_PATH}"

# ------------------------------------------------------------
# Load CSV Files
# ------------------------------------------------------------
train_df = pd.read_csv(TRAIN_PATH)
valid_df = pd.read_csv(VALID_PATH)

print("\n" + "=" * 70)
print("Dataset Loaded Successfully")
print("=" * 70)

print(f"Training Samples   : {len(train_df):,}")
print(f"Validation Samples : {len(valid_df):,}")

display(train_df.head())

# ------------------------------------------------------------
# Convert to HuggingFace Dataset
# ------------------------------------------------------------
train_dataset = Dataset.from_pandas(train_df, preserve_index=False)
valid_dataset = Dataset.from_pandas(valid_df, preserve_index=False)

print("\nConversion Successful")
print(train_dataset)
print(valid_dataset)
```

```python
# ============================================================
# Task 3F - Format Dataset for Phi-4 Training
# ============================================================

def format_example(example):
    text = (
        f"<|user|>\n"
        f"{example['instruction']}\n\n"
        f"Hindi Meaning:\n{example['input']}\n"
        f"<|end|>\n"
        f"<|assistant|>\n"
        f"{example['output']}\n"
        f"<|end|>"
    )

    return {"text": text}


train_dataset = train_dataset.map(format_example)
valid_dataset = valid_dataset.map(format_example)

print("=" * 70)
print("Dataset Formatted Successfully")
print("=" * 70)

print(train_dataset)

print("\nSample Training Example:\n")
print(train_dataset[0]["text"])
```

```python
# ============================================================
# Task 3G - Configure the Trainer
# ============================================================

from pathlib import Path
from transformers import TrainingArguments

# ------------------------------------------------------------
# Project Directories
# ------------------------------------------------------------
PROJECT_DIR = Path("/content/drive/MyDrive/Sanskrit_Phi4_Finetuning")

CHECKPOINT_DIR = PROJECT_DIR / "models" / "phi4_lora_checkpoints"
LOG_DIR = PROJECT_DIR / "logs" / "tensorboard"

CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)
LOG_DIR.mkdir(parents=True, exist_ok=True)

training_args = TrainingArguments(
    output_dir=str(CHECKPOINT_DIR),

    num_train_epochs=3,

    per_device_train_batch_size=1,
    per_device_eval_batch_size=1,

    gradient_accumulation_steps=16,

    learning_rate=2e-4,
    weight_decay=0.01,

    logging_dir=str(LOG_DIR),
    logging_strategy="steps",
    logging_steps=25,

    eval_strategy="epoch",
    save_strategy="epoch",

    load_best_model_at_end=True,
    metric_for_best_model="eval_loss",
    greater_is_better=False,

    save_total_limit=2,

    fp16=True,

    report_to="tensorboard",

    seed=42,

    remove_unused_columns=False,
)

print("=" * 70)
print("TrainingArguments Created Successfully")
print("=" * 70)

print(training_args)
```

```python
# ============================================================
# Task 3H - Create the SFT Trainer
# ============================================================

from trl import SFTTrainer

def formatting_func(example):
    return example["text"]

trainer = SFTTrainer(
    model=model,                     # Already a PeftModel
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=valid_dataset,
    processing_class=tokenizer,
    formatting_func=formatting_func,
)

print("=" * 70)
print("SFT Trainer Created Successfully")
print("=" * 70)

print(type(trainer))
```

```python
from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
PROJECT_DIR = Path("/content/drive/MyDrive/Sanskrit_Phi4_Finetuning")
SPLIT_DIR  = PROJECT_DIR / "outputs" / "dataset_splits"
CONFIG_DIR = PROJECT_DIR / "outputs" / "config"
MODEL_DIR  = PROJECT_DIR / "models"
OUTPUT_DIR = PROJECT_DIR / "outputs" / "generation_results"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

for f in ["train.csv", "validation.csv", "test.csv"]:
    assert (SPLIT_DIR / f).exists(), f"Missing {SPLIT_DIR/f}"
for f in ["lora_config.json", "seed_config.json", "training_config.json"]:
    assert (CONFIG_DIR / f).exists(), f"Missing {CONFIG_DIR/f}"
print("All paths verified OK.")
```

```python
import json, random, os
import numpy as np

with open(CONFIG_DIR / "seed_config.json") as f: seed_cfg = json.load(f)
with open(CONFIG_DIR / "lora_config.json") as f: lora_cfg = json.load(f)
with open(CONFIG_DIR / "training_config.json") as f: train_cfg = json.load(f)

SEED = seed_cfg["seed"]
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
torch.cuda.manual_seed(SEED); torch.cuda.manual_seed_all(SEED)
os.environ["PYTHONHASHSEED"] = str(SEED)

# --- SMOKE TEST TOGGLES ---
# will Flip USE_FULL_DATASET to True once you have flash-attn server access and want
# the real, full-scale run instead of this verification pass.
USE_FULL_DATASET = False
SMOKE_TRAIN_SIZE = 300
SMOKE_VAL_SIZE = 60
SMOKE_EPOCHS = 2

EVAL_STRATEGY = "epoch"
SAVE_STRATEGY = "epoch"
PER_DEVICE_BATCH = 1
GRAD_ACCUM = train_cfg["gradient_accumulation_steps"] * train_cfg["train_batch_size"] // PER_DEVICE_BATCH
print(f"Effective batch size: {PER_DEVICE_BATCH} x {GRAD_ACCUM} = {PER_DEVICE_BATCH*GRAD_ACCUM}")
print("Mode:", "FULL DATASET" if USE_FULL_DATASET else f"SMOKE TEST ({SMOKE_TRAIN_SIZE} train / {SMOKE_VAL_SIZE} val rows, {SMOKE_EPOCHS} epochs)")
```

```python
#LOAD DATASET
#by mistake i re-run the cell thts why the error came
import pandas as pd

train_df_full = pd.read_csv(SPLIT_DIR / "train.csv")
val_df_full   = pd.read_csv(SPLIT_DIR / "validation.csv")
test_df       = pd.read_csv(SPLIT_DIR / "test.csv")

if USE_FULL_DATASET:
    train_df = train_df_full
    val_df = val_df_full
else:
    train_df = train_df_full.sample(n=SMOKE_TRAIN_SIZE, random_state=SEED).reset_index(drop=True)
    val_df = val_df_full.sample(n=SMOKE_VAL_SIZE, random_state=SEED).reset_index(drop=True)

print(f"Train: {len(train_df)}  Val: {len(val_df)}  Test: {len(test_df)}")

N_EVAL_SAMPLES = 20 if not USE_FULL_DATASET else 100
eval_df = test_df.sample(n=min(N_EVAL_SAMPLES, len(test_df)), random_state=SEED).reset_index(drop=True)
few_shot_examples = train_df_full.sample(n=3, random_state=SEED)[["instruction", "output"]].to_dict("records")
print(f"Eval samples: {len(eval_df)}")
```

```python
from unsloth import FastLanguageModel

MODEL_NAME = "unsloth/Phi-4-mini-instruct-unsloth-bnb-4bit"
MAX_SEQ_LENGTH = train_cfg["max_seq_length"]

def load_base_model():
    m, tok = FastLanguageModel.from_pretrained(
        model_name=MODEL_NAME,
        max_seq_length=MAX_SEQ_LENGTH,
        dtype=None,
        load_in_4bit=True,
    )
    return m, tok

model, tokenizer = load_base_model()
print("Model loaded.")
```

```python
#Few-shot examples
#-----------------

@torch.no_grad()
def generate_shloka(model, instruction_prompt, few_shot=None, max_new_tokens=100):
    messages = []
    if few_shot:
        for ex in few_shot:
            messages.append({"role": "user", "content": ex["instruction"]})
            messages.append({"role": "assistant", "content": ex["output"]})
    messages.append({"role": "user", "content": instruction_prompt})

    input_ids = tokenizer.apply_chat_template(
        messages, add_generation_prompt=True, return_tensors="pt"
    ).to(model.device)

    out = model.generate(
        input_ids, max_new_tokens=max_new_tokens, do_sample=False,
        pad_token_id=tokenizer.pad_token_id or tokenizer.eos_token_id,
    )
    return tokenizer.decode(out[0][input_ids.shape[1]:], skip_special_tokens=True).strip()

def generate_for_dataframe(model, df, few_shot=None, max_new_tokens=100):
    results = []
    for i, row in enumerate(df.itertuples(), 1):
        results.append(generate_shloka(model, row.instruction, few_shot, max_new_tokens))
        if i % 10 == 0:
            print(f"  generated {i}/{len(df)}")
    out = df.copy()
    out["generated"] = results
    return out
```

```python
FastLanguageModel.for_inference(model)

print("Baseline ZERO-SHOT...")
baseline_zero_df = generate_for_dataframe(model, eval_df, few_shot=None)
baseline_zero_df.to_csv(OUTPUT_DIR / "baseline_zero_shot.csv", index=False)

print("Baseline 3-SHOT...")
baseline_three_df = generate_for_dataframe(model, eval_df, few_shot=few_shot_examples)
baseline_three_df.to_csv(OUTPUT_DIR / "baseline_three_shot.csv", index=False)
print("Saved.")
```

```python
import re
from skrutable.meter_identification import MeterIdentifier
from sentence_transformers import SentenceTransformer

VALID_ANUSTUBH_PATTERN = re.compile(
    r'^anuṣṭubh \(\d,\d: (?!asamīcīna).+?[;,] \d,\d: (?!asamīcīna).+?\)$'
)
meter_checker = MeterIdentifier()
similarity_model = SentenceTransformer("sanganaka/bge-m3-sanskritFT")

def check_one_verse(verse_text):
    try:
        meter_info = meter_checker.identify_meter(verse_text, from_scheme="DEV")
    except Exception:
        return 0, False
    syllable_count = len(meter_info.syllable_weights.replace("\n", ""))
    is_full_anustubh = bool(VALID_ANUSTUBH_PATTERN.match(meter_info.meter_label))
    return syllable_count, is_full_anustubh

def evaluate_generations(df, label=""):
    syllable_counts, full_flags = [], []
    for verse in df["generated"]:
        n_syl, is_full = check_one_verse(verse)
        syllable_counts.append(n_syl)
        full_flags.append(is_full)

    n = len(df)
    full_pct = 100 * sum(full_flags) / n
    half_pct = max(100 * sum(c == 32 for c in syllable_counts) / n, full_pct)

    meaning_embs = similarity_model.encode(df["input"].tolist(), convert_to_tensor=True)
    gen_embs = similarity_model.encode(df["generated"].tolist(), convert_to_tensor=True)
    cosine_scores = similarity_model.similarity_pairwise(meaning_embs, gen_embs)
    similarity_pct = ((torch.pi - cosine_scores.arccos()) * 100 / torch.pi).mean().item()

    return {"run": label, "n_samples": n, "full_anustubh_pct": full_pct,
            "half_anustubh_pct": half_pct, "semantic_similarity_pct": similarity_pct}
```

```python
baseline_zero_result  = evaluate_generations(baseline_zero_df,  "baseline_zero_shot")
baseline_three_result = evaluate_generations(baseline_three_df, "baseline_three_shot")
pd.DataFrame([baseline_zero_result, baseline_three_result])
```

```python
del model
torch.cuda.empty_cache()

model, tokenizer = load_base_model()

model = FastLanguageModel.get_peft_model(
    model,
    r=lora_cfg["r"],
    lora_alpha=lora_cfg["lora_alpha"],
    lora_dropout=lora_cfg["lora_dropout"],
    bias=lora_cfg["bias"],
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj",
                     "gate_proj", "up_proj", "down_proj"],
    use_gradient_checkpointing="unsloth",
    random_state=SEED,
)
model.print_trainable_parameters()
```

```python
from datasets import Dataset

def tokenize_row(row):
    prompt_msgs = [{"role": "user", "content": row["instruction"]}]
    full_msgs = prompt_msgs + [{"role": "assistant", "content": row["output"]}]
    prompt_ids = tokenizer.apply_chat_template(prompt_msgs, add_generation_prompt=True)
    full_ids = tokenizer.apply_chat_template(full_msgs, add_generation_prompt=False)[:MAX_SEQ_LENGTH]
    labels = list(full_ids)
    for i in range(min(len(prompt_ids), len(full_ids))):
        labels[i] = -100
    return {"input_ids": full_ids, "attention_mask": [1]*len(full_ids), "labels": labels}

train_dataset = Dataset.from_pandas(train_df[["instruction","output"]], preserve_index=False).map(
    tokenize_row, remove_columns=["instruction","output"])
val_dataset = Dataset.from_pandas(val_df[["instruction","output"]], preserve_index=False).map(
    tokenize_row, remove_columns=["instruction","output"])

def collate_fn(batch):
    max_len = max(len(x["input_ids"]) for x in batch)
    pad_id = tokenizer.pad_token_id or tokenizer.eos_token_id
    input_ids, attention_mask, labels = [], [], []
    for x in batch:
        pad_len = max_len - len(x["input_ids"])
        input_ids.append(x["input_ids"] + [pad_id]*pad_len)
        attention_mask.append(x["attention_mask"] + [0]*pad_len)
        labels.append(x["labels"] + [-100]*pad_len)
    return {"input_ids": torch.tensor(input_ids),
            "attention_mask": torch.tensor(attention_mask),
            "labels": torch.tensor(labels)}

print(f"Tokenized. Train: {len(train_dataset)}  Val: {len(val_dataset)}")
```

```python
from transformers import TrainingArguments, Trainer

FastLanguageModel.for_training(model)
CHECKPOINT_DIR = MODEL_DIR / "phi4_lora_checkpoints_smoketest"

training_args = TrainingArguments(
    output_dir=str(CHECKPOINT_DIR),
    num_train_epochs=SMOKE_EPOCHS if not USE_FULL_DATASET else train_cfg["num_train_epochs"],
    per_device_train_batch_size=PER_DEVICE_BATCH,
    per_device_eval_batch_size=PER_DEVICE_BATCH,
    gradient_accumulation_steps=GRAD_ACCUM,
    learning_rate=train_cfg["learning_rate"],
    weight_decay=train_cfg["weight_decay"],
    warmup_ratio=train_cfg["warmup_ratio"],
    lr_scheduler_type=train_cfg["lr_scheduler_type"],
    logging_steps=5,
    eval_strategy=EVAL_STRATEGY,
    save_strategy=SAVE_STRATEGY,
    load_best_model_at_end=train_cfg["load_best_model_at_end"],
    metric_for_best_model=train_cfg["metric_for_best_model"],
    greater_is_better=train_cfg["greater_is_better"],
    save_total_limit=2,
    fp16=not torch.cuda.is_bf16_supported(),
    bf16=torch.cuda.is_bf16_supported(),
    report_to="none",
    seed=SEED,
)

trainer = Trainer(
    model=model, args=training_args,
    train_dataset=train_dataset, eval_dataset=val_dataset,
    data_collator=collate_fn,
)
trainer.train()
```

```python
#SAVING THE TRAINING
#-------------------

SUFFIX = "_smoketest" if not USE_FULL_DATASET else "_final"
FINAL_ADAPTER_DIR = MODEL_DIR / f"phi4_lora{SUFFIX}"
trainer.save_model(str(FINAL_ADAPTER_DIR))
tokenizer.save_pretrained(str(FINAL_ADAPTER_DIR))
print("Saved to", FINAL_ADAPTER_DIR)
```

```python
FastLanguageModel.for_inference(model)

print("Trained ZERO-SHOT...")
trained_zero_df = generate_for_dataframe(model, eval_df, few_shot=None)
trained_zero_df.to_csv(OUTPUT_DIR / f"trained_zero_shot{SUFFIX}.csv", index=False)

print("Trained 3-SHOT...")
trained_three_df = generate_for_dataframe(model, eval_df, few_shot=few_shot_examples)
trained_three_df.to_csv(OUTPUT_DIR / f"trained_three_shot{SUFFIX}.csv", index=False)
print("Saved.")
```

```python
trained_zero_result  = evaluate_generations(trained_zero_df,  "trained_zero_shot")
trained_three_result = evaluate_generations(trained_three_df, "trained_three_shot")

comparison_df = pd.DataFrame([
    baseline_zero_result, baseline_three_result,
    trained_zero_result, trained_three_result,
])
comparison_df.to_csv(PROJECT_DIR / "outputs" / f"comparison{SUFFIX}.csv", index=False)
print(f"--- Mode: {'FULL' if USE_FULL_DATASET else 'SMOKE TEST'} ---")
print(comparison_df.to_string(index=False))
```
